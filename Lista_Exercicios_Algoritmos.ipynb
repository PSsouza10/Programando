## Exercício 1: Conversor de Celsius para Kelvin
Algoritmo que lê um valor de temperatura em graus Celsius, calcula e exibe a temperatura equivalente em graus Kelvin, sabendo que $K = C + 273$.

```python
print("--- 1. Conversor de Celsius para Kelvin ---")
# Entrada de dados
celsius = float(input("Digite a temperatura em graus Celsius (°C): "))

# Processamento
kelvin = celsius + 273

# Saída de dados
print(f"Resultado: {celsius}°C equivale a {kelvin} K.\n")
```

## Exercício 2: Conversor de Polegadas para Milímetros
Algoritmo que lê um comprimento em polegadas, calcula e exibe o comprimento equivalente em milímetros, considerando que 1 polegada equivale a 25,4 milímetros[cite: 1].

```python
print("--- 2. Conversor de Polegadas para Milímetros ---")
# Entrada de dados[cite: 1]
polegadas = float(input("Digite o comprimento em polegadas (\"): "))

# Processamento[cite: 1]
milimetros = polegadas * 25.4

# Saída de dados[cite: 1]
print(f"Resultado: {polegadas}\" equivalem a {milimetros:.2f} mm.\n")
```

## Exercício 3: Cálculo de Potência Mecânica
Algoritmo que utiliza a força F aplicada sobre um corpo e a velocidade V do corpo para calcular e exibir a potência P, utilizando a fórmula $P = F * V$[cite: 1].

```python
print("--- 3. Cálculo de Potência Mecânica ---")
# Entrada de dados[cite: 1]
forca = float(input("Digite a força F aplicada (em Newtons): "))
velocidade = float(input("Digite a velocidade V do corpo (em m/s): "))

# Processamento[cite: 1]
potencia = forca * velocidade

# Saída de dados[cite: 1]
print(f"Resultado: A potência P calculada é de {potencia:.2f} W.\n")
```
